# Лаба 4. A/B/C по CSV


```python
import csv
import json
import statistics
from pathlib import Path

p = Path('docs/experiments')
if not p.is_dir():
    p = Path.cwd()
report = json.loads((p / 'lab4-summary.json').read_text())
for variant in ('quote', 'segment', 'short'):
    with (p / f'lab4-{variant}.csv').open(encoding='utf-8') as stream:
        rows = list(csv.DictReader(stream))
    print(variant, 'F1', report[variant]['f1'],
          'mean generated tokens', statistics.mean(int(r['output_tokens']) for r in rows))
```
