# Лаба 3. Сравнение моделей
Пересчитать метрики по двум CSV. Для повторного вызова моделей используйте lab3.py.


```python
import csv
import sys
from pathlib import Path

location = Path.cwd()
if not (location / 'lab3.py').exists():
    location /= 'docs/experiments'
sys.path.insert(0, str(location))
from lab1 import evaluate

for suffix in ('06b', '4b'):
    with (location / f'lab3-qwen{suffix}.csv').open(encoding='utf-8') as stream:
        rows = list(csv.DictReader(stream))
    metrics = evaluate({r['case']: set(filter(None, r['predicted'].split(','))) for r in rows})
    print(suffix, 'F1', metrics['f1'], 'p95 ms', max(float(r['latency_ms']) for r in rows))
```
